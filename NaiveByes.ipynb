## Naive Baye's Machine Learning Algorithms Implementation

```python
from sklearn.datasets import load_iris 
```

```python
from sklearn.model_selection import train_test_split
X, y = load_iris(return_X_y=True)
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
 
```

## Tips Dataset

```python
import seaborn as sns
```

```python
df=sns.load_dataset("tips")
```

```python
df.head()
```

```python
df['size'].unique()
```

```python
df.isnull().sum()
```

```python
df['sex']=df['sex'].map({'Male':0,'Female':1})
```

```python
df['smoker']=df['smoker'].map({'Yes':0,'No':1})
```

```python
df.head()
```

```python
df['day'].unique()
```

```python
import pandas as pd
```

```python
df['time'].unique()
```

```python
df['time']=df['time'].map({'Lunch':0,'Dinner':1})
```

```python
# One Hot Encoding For Day
day=pd.get_dummies(df['day'],drop_first=True).astype(int)
```

```python
day
```

```python
newDf=pd.concat([df,day],axis=1)
```

```python
newDf
```

```python
newDf.drop(['day'],inplace=True,axis=1)
```

```python
newDf.head()
```

```python
X=newDf.drop(['smoker'],axis=1)
y=newDf['smoker']
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
model_2=GaussianNB()
model_2.fit(X_train,y_train)
y_pred_2=model_2.predict(X_test)
y_pred_2
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test, y_pred_2))
print(classification_report(y_test, y_pred_2))
print(confusion_matrix(y_test, y_pred_2))
```
